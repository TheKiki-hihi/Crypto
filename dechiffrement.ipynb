```python

import os

def déchiffrement_cesar(chaine):
    ListeC=list(chaine)
    D_carac={}
    for i in range(len(ListeC)):
        if ListeC[i] in D_carac :
            D_carac[ListeC[i]]+=1
        else:
            D_carac[ListeC[i]]=1
    nb=0 
    carac=None
    for k,v in D_carac.items():
        if v > nb:
            nb=v
            carac=k
    decalage=(ord(carac)-ord(' '))
    for i in range(len(ListeC)):
        ListeC[i]=chr(ord(ListeC[i])-decalage)
    return "".join(ListeC)
        
with open('messages/message2.txt', encoding='utf-8') as file:
    M2 = file.read()
with open('messages/message3.txt', encoding='utf-8') as file:
    M3 = file.read()

print(déchiffrement_cesar(M2))
print(déchiffrement_cesar(M3))
```

```python

def dechiffrement_pair_impaire(chaine):
    ListeC=list(chaine)
    ListeP=[]
    ListeI=[]
    for i in range(len(ListeC)):
        if i%2!=0:
            ListeI.append(ListeC[i])
        else:
            ListeP.append(ListeC[i])
    chaineI="".join(ListeI)
    chaineP="".join(ListeP)
    chainePD=déchiffrement_cesar(chaineP)
    chaineID=déchiffrement_cesar(chaineI)
    ListeID=list(chaineID)
    ListePD=list(chainePD)
    ListeCD=[]
    for i in range(len(chaineID)):
        ListeCD.append(ListePD[i])
        ListeCD.append(ListeID[i])
    return "".join(ListeCD)

with open('messages/message4.txt', encoding='utf-8') as file:
    M4 = file.read()

print(dechiffrement_pair_impaire(M4))
```
